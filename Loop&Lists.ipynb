# For Loop

```python
# For Loop
numbers = range(5)
for i in numbers:
  print(i)
```

```python
# Simple For Loop
for i in range(1,10):
  print(i)
```

```python
# For loop step increment
for i in range(1,10,2):
  print(i)
```

```python
# While loop
i = 1
while i <= 10:
  print(i)
  i = i + 1
```

# Tuple


## Definition

### A tuple is ordered collection of items
### It is immutable
### Tuple given in the ()

```python
# Create a Tuple
student = ("Chirag",21,'CSE',8.7)
print(student)
```

```python
# display the items by index
subjects=("python","DSA","DAA","OS","SQL","Statistics")
print(subjects[0])
print(subjects[5])
```

```python
subjects=("python","DSA","DAA","OS","SQL","Statistics")
subjects[2]="Algebra"
```

```python
# Tuple slicing and print the length,find the item is present in tuple
subjects=("Python",
          'Statistics',
          'SQL',"Machine Learning","Deep Learning")
print(subjects[1:4])
print("Number of Subjects:",len(subjects))
print("Deep Learning" in subjects)
print("Relational Algebra" in subjects)
```

```python
# Nested tuple
student=(101,'Rahul',(85,98,76))
print(student)
print(student[2])
print(student[2][1])
```
