#Tuple
## Ordered collection of items ()
## One main difference with is tuple is immutable
## var_name = list[[stu1:()],[stu2:()]....]


```python
# Create a tuple
student = (101,"Sanjiv","CSE",8.9)
print(student)

# tuples using index
print("Student Information in first location:",student[0])
print(student[1])

# Negative index
print("Example for negative index:",student[-2])

```

```python
# to find length of the tuple
subjects=("DSA",'DSV',"OS","Python","Machine Learning")
print(len(subjects))
```

```python
# Slicing
subjects=("DSV",'DSA',"Statistics","SQL","Machine Learning")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
# tuple immutable
students_info=("Chirag",21,"CSE",8.7)
students_info[3] = 9.1
print(students_info)
```

```python
#Nested tuple
student=(101,"Rahul",(89,78,66))
print(student)
print(student[2])
print(student[2][1])
```

```python
# Count function
marks = (75,78,68,75,89,75,92,98)
print("Time of Occurance:",marks.count(75))
```

```python
# find the element present in tuple or not
subjects=("Python",
          "DSV",
          "SQL",
          "Machine Learning",
          "Statistics")
print("Python" in subjects)
print("Java" in subjects)
```

# SET
## unordered collection of items
## mutable
## allows the different data types
## it doesn't support for index
## it gives output with unoique values(No duplication)

```python
# Symbol for set { }
# Create a set with similar items
marks={75,68,89,55,78,89,75,66,55,99}
print(marks)
print(type(marks))
```

```python
# Create a empty set
a={}
a=set()
# check its type
print(type(a))
```

```python
# set doesn't support the index
students={"Priyeranjan","Sakshi","Shivam"}
print(students[1])
```

```python
# add an item in the set
students={"Priyeranjan","Sakshi","Shivam"}
students.add("Riya")
print(students)
# to add multiple values or items
students.update(["Ram","Chitra","Vani","Devi"])
print(students)
```

```python
# remove an item from the set
students={"Priyeranjan","Sakshi","Shivam"}
students.remove("Shivam")
print(students)

students.discard("Priyeranjan")
print(students)

```

```python
students={"Priyeranjan","Sakshi","Shivam"}
students.remove("Karan")
print(students)
```

```python
students={"Priyeranjan","Sakshi","Shivam"}
students.discard("Karan")
print(students)
```

```python
# create a set and clear list of items
students_info={"Priyeranjan",
               "Sakshi",
               "Shivam"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks={99,88,77,66}
del marks
print(marks)
```

```python
# Set Operations
python_students={"Priyeranjan","Sakshi","Shivam","Riya"}
java_students={"Karan","Vansh","Yuvraj","Piyush","Neha"}

#for union operation by using set
result=python_students.union(java_students)
print("Union Operation:",result)

#for intersection operation by using set
result_intersection=python_students.intersection(java_students)
print("Intersection Operation:",result_intersection)

# to find difference with two set
res_diff=python_students.difference(java_students)
print("For Difference:",res_diff)
```

```python
# Convert list to set
stud_id=[101,102,103,102,103,104,105]
unique_stud_id=set(stud_id)
print(unique_stud_id)
uniq_roll_no=list(unique_stud_id)
print(uniq_roll_no)
```
